```python
import os
from getpass import getpass
from pydantic import BaseModel, Field
from langchain_openai import ChatOpenAI

os.environ["OPENAI_API_KEY"] = getpass("OPENAI_API_KEY: ")

llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0,
)
```

```python
response = llm.invoke(
    "Alice는 30살입니다. 이 사람을 설명해줘."
)

print(type(response))
print(response.content.replace('.', '.\n'))
```

```python
class Person(BaseModel):
    name: str = Field(description="사람의 이름")
    age: int = Field(description="사람의 나이")


structured_llm = llm.with_structured_output(Person)

result = structured_llm.invoke(
    "Alice는 30살입니다. 이 사람을 설명해줘."
)

print(type(result))
print(result)
print(result.name)
print(result.age)
```

로그 예시

```python
from datetime import datetime, timedelta
import random

levels = ["INFO", "WARNING", "ERROR", "DEBUG"]
services = ["auth", "payment", "order", "user", "notification"]

messages = [
    "User login successful",
    "Invalid password attempt",
    "Payment processed",
    "Payment failed",
    "Order created",
    "Order cancelled",
    "User profile updated",
    "Email notification sent",
    "Database connection retry",
    "API response time exceeded threshold",
]

base_time = datetime.now()

fake_logs = []

for i in range(20):
    log_time = base_time - timedelta(minutes=i * random.randint(1, 5))
    log = {
        "timestamp": log_time.strftime("%Y-%m-%d %H:%M:%S"),
        "level": random.choice(levels),
        "service": random.choice(services),
        "message": random.choice(messages),
    }
    fake_logs.append(log)

for log in fake_logs:
    print(
        f"[{log['timestamp']}] "
        f"{log['level']} "
        f"service={log['service']} "
        f"message=\"{log['message']}\""
    )
```

```python
import re
import pandas as pd

# fake_logs를 문자열 로그로 변환
log_lines = [
    f"[{log['timestamp']}] {log['level']} service={log['service']} message=\"{log['message']}\""
    for log in fake_logs
]

# 로그 파싱 패턴
pattern = re.compile(
    r'\[(?P<timestamp>.*?)\]\s+'
    r'(?P<level>\w+)\s+'
    r'service=(?P<service>\w+)\s+'
    r'message="(?P<message>.*?)"'
)

parsed_logs = []

for line in log_lines:
    match = pattern.match(line)
    if match:
        parsed_logs.append(match.groupdict())

# DataFrame으로 보기 좋게 출력
df_logs = pd.DataFrame(parsed_logs)

display(df_logs)
```
